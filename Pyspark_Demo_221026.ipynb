This workbook is a Pyspark demo that consists of preprocessing a Sales dataset. 

```python
import pyspark as pyspark
import time as time 
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("Spark App").getOrCreate()

sales = spark.table("pyspark_exercise.default.big_mart_sales")
sales.show(3)
```

```python
sales.printSchema()
sales.show(15, truncate=False)
```

```python
sales_string = sales.withColumn("Item_Weight_new", col("Item_Weight").cast("string")).drop(sales.Item_Weight)
sales_string = sales_string.withColumnRenamed("Item_Weight_new", "Item_Weight")
sales_string.printSchema()
sales_string.show(3)
```
