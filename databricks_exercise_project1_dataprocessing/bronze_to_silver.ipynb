```python
from pyspark.sql import functions as F

# Specify Data types 
schema = 'user_id INT, user_name STRING, signup_date STRING'

# Read the dirty data
df = spark.table("etl_exercise_project_1_catalog.data1.users_dirty_csv")

# Step 1: Remove duplicate user_ids (keeps first occurrence, removes latter duplicates)
df_deduped = df.dropDuplicates(subset=["user_id"])

# Step 2: Standardize date format - replace dots with slashes (M.D.YY → M/D/YY)
df_cleaned = df_deduped.withColumn(
    "signup_date",
    F.regexp_replace("signup_date", r"\.", "/")
)
```

```python
df_cleaned.write.mode("overwrite").saveAsTable("etl_exercise_project_1_catalog.data1.users_cleaned_csv")
```
